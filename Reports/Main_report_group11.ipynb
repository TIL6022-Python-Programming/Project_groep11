# 1. Introduction
The aviation industry is vulnerable to external disruptions, such as geopolitical events, economic crises, and pandemics. These events can have a major impact on passenger numbers and airport operations. However, the way passenger traffic responds can vary significantly depending on the event. 

It is important for airports to gain insight into these differences. A better understanding of the impact of and recovery from past disruptions can help them prepare for future events. This is relevant for capacity planning and recovery strategies, for example. By analyzing historical trends, patterns can be identified that provide insight into the resilience of passenger traffic.

This study focuses on Amsterdam Airport Schiphol and compares the effects of three different types of external disruptions: the 9/11 attacks of 2001, the global financial crisis of 2008–2009, and the COVID-19 pandemic. These events were selected because they differ in cause, duration, and the way in which they can affect aviation. 

# 2. Research objective & questions
For this research the following research question has been formulated, supported by three subquestions. 

## Research Question
How do different major external disruptions affect passenger traffic at Amsterdam Airport Schiphol, and what insights do these effects reveal about airport resilience?

## Subquestions
**SQ1. How do the magnitude, duration, and recovery patterns of changes in international passenger traffic differ across external disruptions?**

**SQ2. How are the characteristics of external disruptions related to differences in their impact on international passenger traffic?**

**SQ3. What can these differences in impact and recovery patterns reveal about the resilience of international passenger traffic at Schiphol Airport?**
