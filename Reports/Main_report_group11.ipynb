# 1. Introduction
The aviation industry is vulnerable to external disruptions, such as geopolitical events, economic crises, and pandemics. These events can have a major impact on passenger numbers and airport operations. However, the way passenger traffic responds can vary significantly depending on the event. 

It is important for airports to gain insight into these differences. A better understanding of the impact of and recovery from past disruptions can help them prepare for future events. This is relevant for capacity planning and recovery strategies, for example. By analyzing historical trends, patterns can be identified that provide insight into the resilience of passenger traffic.

This study focuses on Amsterdam Airport Schiphol and compares the effects of three different types of external disruptions: the 9/11 attacks of 2001, the global financial crisis of 2008–2009, and the COVID-19 pandemic. These events were selected because they differ in cause, duration, and the way in which they can affect aviation.

# 3. Characteristics of external disruptions

External disruptions have varying characteristics and can therefore affect passenger traffic at Amsterdam Schiphol in different ways. The three disruptions analysed in this report differ in nature and progression. While one event arises suddenly and primarily affects passengers’ perception of safety, another develops over a longer period and limits the financial ability to travel. In addition, government measures can also directly limit the ability to travel internationally.

This chapter discusses the backgrounds and different characteristics of these disruptions, focusing on their origins, progression, and potential consequences for passenger traffic at Amsterdam Schiphol.

## 9/11 attacks
On September 11, 2001, four passenger planes were hijacked in the United States. Two planes were flown into the Twin Towers of the World Trade Center in New York, causing both towers to collapse. A third plane struck the Pentagon near Washington, D.C. The fourth plane crashed in Pennsylvania. Nearly three thousand people lost their lives in the attacks.

9/11 was a sudden, security-related disruption that directly targeted civil aviation. Although the attacks took place in the United States, additional security measures were introduced internationally. A key characteristic of these attacks was the contrast between the brief duration of the attacks themselves and the lasting changes in aviation security that followed.

Security measures were also tightened at Schiphol airport. Passengers flying to the US were subject to additional screenings, and extra screening checkpoints were added. Overall security measures were also tightened at Schiphol airport. These included stricter ticket and passport checks, tighter access controls, and more intensive aircraft inspections.

The attacks also affected passengers' sense of security. Passengers felt more anxious and began to associate flying with a greater risk. In addition, the stricter security checks made traveling less convenient and comfortable.

The key characteristics of 9/11 that are relevant to this study are its sudden onset, its direct connection to aviation security, and its international consequences in the form of additional security measures and concerns about safety.

## COVID-19 pandemic
The COVID-19 pandemic was a global health crisis caused by the SARS-CoV-2 coronavirus. After the first reported outbreak in late 2019, the virus spread internationally. In March 2020, the first measures to control the virus were implemented in the Netherlands. These measures had a significant impact on international travel and were therefore highly relevant to air traffic to and from Schiphol.

Unlike a one-time event, the COVID situation was constantly changing due to successive waves of infection and new virus variants. As a result, restrictions could be eased and then tightened again later. The measures and the timing of their implementation also varied from country to country.

In the Netherlands, various measures were implemented to limit the spread of the coronavirus through international travel. These included temporary flight bans for certain countries, an EU entry ban, and testing and quarantine requirements for certain groups of travelers.

In addition, international air travel depended on measures in both the country of departure and the country of destination. An easing of restrictions in the Netherlands therefore did not automatically mean that all international travel was once again possible without restrictions.

In addition to these measures and restrictions, concerns about infection and uncertainty about changing travel conditions could influence the willingness to travel. Travel became less appealing, even though it was technically possible.

The key characteristics of COVID-19 relevant to this study are its global spread, its long duration with multiple waves, and the combination of health concerns and government measures. 

## Global financial crisis 2007-2009

The global financial crisis began in 2007 with problems in the U.S. mortgage market. Financial institutions were weakened by falling home prices and losses on mortgage-related investments. Due to the interconnected nature of the financial system, these problems quickly spread to other countries. Following the bankruptcy of Lehman Brothers in September 2008, the crisis intensified worldwide.

The Netherlands was also affected by this, leading the Dutch government to take measures to support financial institutions. What began as instability in the banking sector developed into a broader economic downturn. Financial losses and difficulties in obtaining credit put households and businesses under pressure, while uncertainty about future economic conditions undermined confidence. The crisis unfolded over a longer period of time, rather than being a one-time, isolated event.

The international aviation sector was also affected by the economic downturn. It may have influenced both the decision to travel and the amount passengers were willing or able to spend. Households facing financial uncertainty may have become more cautious about their travel spending, while companies may have cut back on their travel budgets. The recession’s impact on incomes and confidence was identified as a key factor influencing demand for air travel

The key characteristics of the financial crisis that are relevant to this study are its international scope, its duration over a longer period, and its impact on financial resources and confidence. These characteristics provide the context for the analysis of changes in passenger traffic and the subsequent recovery at Amsterdam Airport Schiphol.

The three disruptions differ in terms of their causes, their progression, and the way in which they can affect passenger traffic. These differences form the basis for the further course of this study. Using passenger data and additional information, the study will examine how traffic at Amsterdam Airport Schiphol evolved around each disruption and what differences are evident in the decline and recovery.

# 5. Methodology & data processing

In this study, we examined how passenger traffic at Schiphol changed in the wake of three different disruptions: the 9/11 attacks, the COVID-19 pandemic and the global financial crisis.

We processed the data in Python using Pandas, and the graphs were created with Matplotlib. The study takes a descriptive and comparative approach, focusing on changes in passenger numbers, the duration of these changes, and the subsequent recovery patterns.

The dataset contains monthly passenger numbers and flight movements at Schiphol. We selected the necessary columns, removed rows that did not represent a calendar month, and combined the year and month into a single date. For September 11, we used the original Excel file because the reference period begins in 1996, whereas the processed CSV file does not begin until 1999. For the financial crisis and COVID-19, we used the CSV file.

## 5.2 Passenger Traffic Indicators
For each disruption, we first created a graph showing the original monthly passenger numbers, which are presented in Figure 5.1. Seasonal fluctuations are still visible in these three graphs. For example, passenger numbers are generally higher in the summer than in the winter. This makes it difficult to determine, based on the raw data, whether a decline is greater than normal for that time of year.


 To account for these seasonal fluctuations, we compared each month to the average for the same calendar month during the reference period. The actual number was then divided by that average and multiplied by 100. An index of 100 means that the number of passengers was at the usual level for that month. In the same way, we calculated an index for the number of flight movements so that we could compare both trends.

In addition, we calculated the percentage change compared to the same month a year earlier. This allowed us to see when passenger numbers began to decline or started to rise again. If the number a year earlier was very low, the growth might appear large. Therefore, we used the historical index to assess when passenger traffic returned to the reference level. 
To get a longer-term picture we also compared the annual totals to 2000 (before 9/11), 2007 (before the financial crisis), and 2019 (before COVID-19) in another graph. For incomplete years, we compared the available months to the same months in the reference year.
For the last analyses, we compared passengers against flight movements, to see wether airlines behaved the same as passengers did (e.g. if plans were fully occupied).

## 5.3 Growth Models

Actual passenger numbers were also compared with forecasts based on past growth. For 9/11 and COVID-19, we calculated the average growth per calendar month compared to the same month a year earlier, over a five-year period. Using those growth rates, we extrapolated the passenger numbers from the last full year before the disruption to the subsequent years. For the financial crisis, we used a rolling five-year average of growth. For each month, we applied this to the actual passenger number for the same month a year earlier.

We expressed the difference between the actual and predicted numbers as a percentage of the predicted number. Because the model for the financial crisis continuously incorporates new months, we based the comparison between the three disruptions primarily on the indicators that we calculated in the same way for all three.

## 5.4 Additional Data and Comparisons

To validate our findings, we made additional graphs. These graphs vary for every event and require different additional sources. For the 9/11-event, we use data from the U.S. Bureau of Transportation Statistics to compare the percentage change on routes between Schiphol and the United States with the change in total passenger traffic at Schiphol, before and after the event. For the financial crisis, we compared the deviation from the growth model with the AEX index to see if both followed a similar trend. With regards to COVID-19, we cross-checked Schiphol’s figures against aviation statistics from Statistics Netherlands (CBS). We used CBS figures on international vacations to put the changes in the number of flight movements into context.

Because the total number of passengers varied by period, we primarily used percentages and index figures to compare the three disruptions. When interpreting the results, we took into account the characteristics of each disruption, as described in Chapter 3.

## 9/11 attacks
When looking at the passenger data before, after and mainly around 9/11, a clear decrease in the passenger numbers can be seen in the year-on-year comparison graph. The biggest decline was seen in October 2001, of -11.6% compared to October in the year before. During 2002, passenger traffic started to recover, with some months showing positive growth again. In 2003, passenger growth became negative again. This was not necessarily caused by 9/11, since other events, such as SARS, the Iraq War and the economic situation, also affected air travel during this period. Overall, the graph shows that 9/11 was followed by a sudden drop in passenger traffic, while the recovery took more time and was also affected by other events. (year on year passenger change)


The passenger numbers were also compared with aircraft movements. In this graph, we can see that both are affected by 9/11, but not in the same way. After the initial drop, passenger numbers generally show a better recovery pattern than aircraft movements. The prediction model shows that passenger traffic drops further below the expected level after 9/11, indicating a big deviation. ## 7.1 Magnitude and development of the disruptions

The largest monthly decline compared to the same month in the previous year during this period was 14.5% in March 2009.

The COVID-19 pandemic led to a massive and more abrupt decline. In April 2020, passenger numbers were 97.9% lower than in April 2019. This was an exceptionally large decline compared to the other two disruptions.

## 7.2 Duration and recovery patterns
When examining the recovery of passenger traffic, differences between the disruptions are clearly visible. Figure 7.2 compares the annual passenger numbers for each disruption with those of the previous year.


![Figure year-over-year change in passengers.png](<attachment:Figure year-over-year change in passengers.png>)

After September 11, passenger numbers began to recover in the course of 2002. In 2002, annual passenger numbers reached approximately 103.4% of the 2000 level. It is interesting to notice that passenger traffic in 2001 was already slightly higher than in 2000, after which, in 2002, it reached 103.4% of the 2000 level. This illustrates how annual totals can mask a disruption that occurs late in the year.

The financial crisis was accompanied by a more prolonged period of declining passenger numbers. Annual passenger traffic in 2009 was approximately 91.2% of the 2007 level. The annual number of passengers then increased, reaching the 2007 reference level in 2011, when it was approximately 104.1%.

The recovery after COVID-19 followed a different pattern. After the massive decline in 2020, passenger numbers gradually increased, particularly as international travel restrictions were eased. However, passenger traffic remained below the pre-pandemic annual level for several years. In 2025, the annual number of passengers was approximately 95.9% of the 2019 level.

## 7.3 Relationship between event characteristics and passenger traffic
The observed differences may be related to the characteristics of the disruptions, as discussed in Chapter 3. Another aspect of the comparison is the relationship between the number of passengers and the number of flight movements, as shown in Figure 7.4.


![Figure passengers versus flight movements.png](<attachment:Figure passengers versus flight movements.png>)
**Figure 7.4:** Comparison of passenger traffic and flight movements relative to their historical reference levels.

The relationship between the number of passengers and the number of flight movements provides additional insight into the trends in air traffic during the disruptions. Although both indicators were affected, they did not necessarily change at the same rate, neither did they recover at the same rate. This shows that changes in passenger traffic and flight activity were not always the same.

The September 11 attacks were a sudden security incident that had a direct impact on the sense of security in aviation. Examining the routes between Schiphol and the U.S. highlights this geographical link. Passenger traffic on these routes in October 2001 fell by approximately 33.6% compared to the same month in the previous year, while the decline for total passenger traffic at Schiphol was 11.6%. This suggests that the disruption affected some parts of Schiphol’s airline network more than others.

For the financial crisis, the situation was different, as the crisis had a broad impact on the economic situation of people and businesses and, therefore, on the travel decisions they made. Financial uncertainty and reduced spendable income may have contributed to the long-term decline in passenger numbers. The comparison with the AEX index provides additional economic context, although no direct causal link can be drawn from this.

COVID-19 led to health concerns and government restrictions on international travel. Unlike a disruption that primarily affects the willingness or financial ability to travel, these restrictions could directly limit travel capabilities. This difference offers a possible explanation for the extremely large and abrupt decline in passenger numbers.

Together, the three cases demonstrate that the characteristics of a disruption can be linked to the magnitude of its impact and the way passenger traffic evolves over time. However, the identified patterns cannot be attributed solely to the selected disruptions, as there are other factors that may have influenced passenger traffic as well.


## 7.4 Implications for airport resilience

The comparison shows that resilience cannot be determined solely by the magnitude of a disruption at the moment it occurs. The duration of the reduced traffic volume, the speed of recovery, and the ability to return to previous passenger levels also provide relevant information.

For airports such as Schiphol, these findings demonstrate how important it is to be prepared for various types of disruptions. Sudden security-related incidents, prolonged economic recessions, and restrictions on international travel may require different forms of capacity planning and recovery strategies. However, passenger traffic is only one aspect of airport resilience.
