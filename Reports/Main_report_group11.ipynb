# 1. Introduction
The aviation industry is vulnerable to external disruptions, such as geopolitical events, economic crises, and pandemics. These events can have a major impact on passenger numbers and airport operations. However, the way passenger traffic responds can vary significantly depending on the event. 

It is important for airports to gain insight into these differences. A better understanding of the impact of and recovery from past disruptions can help them prepare for future events. This is relevant for capacity planning and recovery strategies, for example. By analyzing historical trends, patterns can be identified that provide insight into the resilience of passenger traffic.

This study focuses on Amsterdam Airport Schiphol and compares the effects of three different types of external disruptions: the 9/11 attacks of 2001, the global financial crisis of 2008–2009, and the COVID-19 pandemic. These events were selected because they differ in cause, duration, and the way in which they can affect aviation.

# 3. Characteristics of external disruptions

External disruptions have varying characteristics and can therefore affect passenger traffic at Amsterdam Schiphol in different ways. The three disruptions analysed in this report differ in nature and progression. While one event arises suddenly and primarily affects passengers’ perception of safety, another develops over a longer period and limits the financial ability to travel. In addition, government measures can also directly limit the ability to travel internationally.

This chapter discusses the backgrounds and different characteristics of these disruptions, focusing on their origins, progression, and potential consequences for passenger traffic at Amsterdam Schiphol.

## 9/11 attacks
On September 11, 2001, four passenger planes were hijacked in the United States. Two planes were flown into the Twin Towers of the World Trade Center in New York, causing both towers to collapse. A third plane struck the Pentagon near Washington, D.C. The fourth plane crashed in Pennsylvania. Nearly three thousand people lost their lives in the attacks.

9/11 was a sudden, security-related disruption that directly targeted civil aviation. Although the attacks took place in the United States, additional security measures were introduced internationally. A key characteristic of these attacks was the contrast between the brief duration of the attacks themselves and the lasting changes in aviation security that followed.

Security measures were also tightened at Schiphol airport. Passengers flying to the US were subject to additional screenings, and extra screening checkpoints were added. Overall security measures were also tightened at Schiphol airport. These included stricter ticket and passport checks, tighter access controls, and more intensive aircraft inspections.

The attacks also affected passengers' sense of security. Passengers felt more anxious and began to associate flying with a greater risk. In addition, the stricter security checks made traveling less convenient and comfortable.

The key characteristics of 9/11 that are relevant to this study are its sudden onset, its direct connection to aviation security, and its international consequences in the form of additional security measures and concerns about safety.

## COVID-19 pandemic
The COVID-19 pandemic was a global health crisis caused by the SARS-CoV-2 coronavirus. After the first reported outbreak in late 2019, the virus spread internationally. In March 2020, the first measures to control the virus were implemented in the Netherlands. These measures had a significant impact on international travel and were therefore highly relevant to air traffic to and from Schiphol.

Unlike a one-time event, the COVID situation was constantly changing due to successive waves of infection and new virus variants. As a result, restrictions could be eased and then tightened again later. The measures and the timing of their implementation also varied from country to country.

In the Netherlands, various measures were implemented to limit the spread of the coronavirus through international travel. These included temporary flight bans for certain countries, an EU entry ban, and testing and quarantine requirements for certain groups of travelers.

In addition, international air travel depended on measures in both the country of departure and the country of destination. An easing of restrictions in the Netherlands therefore did not automatically mean that all international travel was once again possible without restrictions.

In addition to these measures and restrictions, concerns about infection and uncertainty about changing travel conditions could influence the willingness to travel. Travel became less appealing, even though it was technically possible.

The key characteristics of COVID-19 relevant to this study are its global spread, its long duration with multiple waves, and the combination of health concerns and government measures. 

## Global financial crisis 2007-2009

The global financial crisis began in 2007 with problems in the U.S. mortgage market. Financial institutions were weakened by falling home prices and losses on mortgage-related investments. Due to the interconnected nature of the financial system, these problems quickly spread to other countries. Following the bankruptcy of Lehman Brothers in September 2008, the crisis intensified worldwide.

The Netherlands was also affected by this, leading the Dutch government to take measures to support financial institutions. What began as instability in the banking sector developed into a broader economic downturn. Financial losses and difficulties in obtaining credit put households and businesses under pressure, while uncertainty about future economic conditions undermined confidence. The crisis unfolded over a longer period of time, rather than being a one-time, isolated event.

The international aviation sector was also affected by the economic downturn. It may have influenced both the decision to travel and the amount passengers were willing or able to spend. Households facing financial uncertainty may have become more cautious about their travel spending, while companies may have cut back on their travel budgets. The recession’s impact on incomes and confidence was identified as a key factor influencing demand for air travel

The key characteristics of the financial crisis that are relevant to this study are its international scope, its duration over a longer period, and its impact on financial resources and confidence. These characteristics provide the context for the analysis of changes in passenger traffic and the subsequent recovery at Amsterdam Airport Schiphol.

The three disruptions differ in terms of their causes, their progression, and the way in which they can affect passenger traffic. These differences form the basis for the further course of this study. Using passenger data and additional information, the study will examine how traffic at Amsterdam Airport Schiphol evolved around each disruption and what differences are evident in the decline and recovery.

# 6.Analysis
## 9/11 attacks
When looking at the passenger data before, after and mainly around 9/11, a clear decrease in the passenger numbers can be seen in the year-on-year comparison graph. The biggest decline was seen in October 2001, of -11.6% compared to October in the year before. During 2002, passenger traffic started to recover, with some months showing positive growth again. In 2003, passenger growth became negative again. This was not necessarily caused by 9/11, since other events, such as SARS, the Iraq War and the economic situation, also affected air travel during this period. Overall, the graph shows that 9/11 was followed by a sudden drop in passenger traffic, while the recovery took more time and was also affected by other events. (year on year passenger change)


The passenger numbers were also compared with aircraft movements. In this graph, we can see that both are affected by 9/11, but not in the same way. After the initial drop, passenger numbers generally show a better recovery pattern than aircraft movements. This suggests that passenger traffic and the number of flights did not recover at the same pace. Possible explanations for this could be that airlines adjusted their flight schedules by decreasing the number of flights or that larger aircrafts were used to transport more passengers with fewer flights. (passenger vs aircraft movements)


## COVID-19
After COVID-19 spread at the end of 2019, the first measures against COVID-19 were implemented in March of 2020. These measures included countries such as Denmark, Poland, Taiwan, Vietnam, etc. closing borders for foreigners. 

When looking at the passengers per month from Schiphol airport, it can be seen that at the end of 2019 - beginning of 2020 the air traffic already started to drop a bit. The real drop however is in April of 2020. Now in this graph it is hard to see the direct impact, therefore each month will be compared to the same calendar month a year before. (passengers per month)

In April of 2020 the shock was most noticeable, with the index hitting 2.3. Indicating that in April of 2020, compared to the baseline of April 2015-2019, only 2.3% passengers flew this period. After this period, where measures were getting less and less strict, it can be seen that passenger numbers went up again. (passengers compared to baseline)

From this graph the recovery rate can be deduced. The graph shows the percentage of passengers compared to the same month in the previous year. Showing decline (73-98%) for the period of April 2020 - March 2021, but also extreme growth from March 2021 onwards (up to +524%, April 2021). The standard year-on-year growth (from before COVID pandemic) was back to normal in 2024, indicating that the actual negative period of the COVID pandemic took 1 year. Whilst the actual recovery of the negative period took a bit less than 3 years. (year over year change in passengers )

When looking at the passengers vs. flight movements. It is interesting to note that there were relatively more flights after the initial hit in April 2020. This was most likely caused by the measures such as 1.5 metres distance, making it mandatory to have an empty seat between passengers, thus making planes less full. From 2022 onwards the lines get closer together, this also being around the time the 1.5 metre distance measure was not mandatory anymore https://www.rijksoverheid.nl/actueel/nieuws/2022/02/24/anderhalve-meter-afstand-niet-meer-verplicht . (passengers versus flight movements)

Obviously no one expected the COVID-19 pandemic to happen, nor to be so disrupting of the life we see as normal. Looking at the predicted growth for Schiphol airport, there can be seen that there is a massive overestimation in hindsight. The growth rate pre-covid was already optimistic, with Schiphol not hitting their targets in the years 2018-2019. Now in 2026 Schiphol is still 25-35% smaller than the expectations. The gap can also be explained by holdback from the government running policy to keep the capacity of Schiphol down https://nltimes.nl/2026/03/24/government-sticks-478000-schiphol-flight-cap-despite-legal-setback . (deviation from the forecast model)


The total picture of the amount of yearly passengers shows the absolute number of passengers compared to the baseline set in 2019. With 2020 being the worst year for Schiphol with only 29% of passengers compared to 2019. After 2020 the amount of passengers grew again, most notably from 2022 when most measures were canceled. In 2024 the numbers were almost back to the baseline level of 2019, with 2026 being predicted to break the baseline level (numbers of 2026 are till August) and thus being fully recovered. (yearly total passengers)

## Validation
To really make sure these numbers are not coincidental and are caused by the COVID-pandemic, we will also look at the total number of vacations abroad of Dutch citizens. 

Flight traffic was directly hit by the measures, since it simply did not allow for planes to take off. However, people could still get in their car and drive to a neighbouring country. This also shows in the data (CBS). The relative number of Dutch citizens going on an abroad vacation exceeds the number of flights, but still sees a huge decline like the flight traffic from Schiphol. (flights from Schiphol and vacations abroad)

Comparing the flights and vacations abroad it is noticeable that the drop in 2020 and 2021 is almost the same. The recovery in 2022 however is much in favor of vacations abroad. Measures being dropped one-by-one allowed people to visit neighbouring countries sooner, thus recovering faster. (change compared to the previous year)

## Global financial crisis
The monthly passenger figures show that growth had already begun to level off in late 2007 and early 2008. The most pronounced decline became apparent in late 2008, as the financial crisis worsened, and continued into 2009. Due to seasonal fluctuations, it is difficult to determine from the graph with absolute figures to what extent passenger traffic deviates from its usual level. For this reason, each month is compared with the same calendar month a year earlier. (Passengers per month)

The seasonally adjusted index shows that the decline was most pronounced in 2009. For an extended period, the index remained well below the 2004–2007 reference level. During that period, Schiphol therefore handled structurally fewer passengers than expected based on that reference level. When economic conditions subsequently began to stabilize gradually and the financial markets recovered, passenger numbers began to rise again. (Passengers compared to baseline )

This graph shows the percentage change in passenger numbers compared to the same month a year earlier. From late 2008 to late 2009, the changes were predominantly negative. After that, passenger numbers began to rise again compared to the previous year. In 2011, year-over-year growth returned to roughly the same pace as before the crisis. The period of decline lasted about a year and a half. It is estimated that it will take three to four years for the cumulative number of passengers to fully recover.

Two notable outliers are April 2010 (−18.1%) and April 2011 (+37.2%). In April 2010, the eruption of the Icelandic volcano Eyjafjallajökull caused an ash cloud over Europe. Parts of European airspace were closed for several days, resulting in the cancellation of many flights at Schiphol and a sharp decline in passenger numbers that month.

The sharp increase in April 2011 is due to a base effect. The number of passengers for that month is compared to the exceptionally low figures from April 2010, resulting in a large percentage increase. (year-over-year change in passengers)


Comparing the number of passengers with the number of flight movements provides insight into how airlines have adjusted their capacity. During the financial crisis, demand for air travel came under pressure, partly because companies cut back on business travel. Airlines responded by grounding aircraft and reducing flight frequencies. As a result, the indices for passenger numbers and flight movements remained relatively close to one another during the recession. (passengers versus flight movements)


The growth model extrapolates the growth trend from the five years preceding the crisis. A comparison with actual passenger numbers shows in hindsight that this growth trend was too high. At its lowest point, passenger traffic was approximately 15 to 20 percent below the level the model had predicted based on pre-crisis growth. (deviation from 5-year growth model)

Annual passenger numbers are compared to 2007, the chosen reference year. In 2009, Schiphol reached its lowest level during the period under review, at 91% of the 2007 passenger volume. After 2009, passenger numbers began to rise again. In 2011, they reached 104% of the reference level, and in 2012, 108%. As a result, annual passenger numbers have been above the 2007 level again since 2011. (Yearly Total Passengers )

## Validation
To determine whether the trend in passenger traffic coincided with the broader economic downturn, we compare the deviation from the growth model with the AEX index. During the crisis, households and businesses cut back on their spending, including travel expenses. At the same time, stock prices fell. The comparison shows that the deviation in passenger traffic and the AEX index followed a similar pattern during the 2008–2009 downturn and the subsequent recovery.(Schiphol Traffic Deviation vs. AEX Stock Market Index)
